```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons

# 1. Generate the raw array data
X, y = make_moons(n_samples=150, noise=0.2, random_state=42)

# 2. Create the pandas DataFrame and label it 'df'
# We assign the two columns in X to 'Feature_1' and 'Feature_2'
df = pd.DataFrame(X, columns=['Feature_1', 'Feature_2'])

# 3. Add the target labels to the DataFrame
df['Label'] = y

# View the first few rows to confirm it worked!
print(df.head())
```

```python
X = df.iloc[:,0:2].values
y = df.iloc[:,-1].values
```

```python
import tensorflow
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense
```

```python
model = Sequential()

model.add(Dense(10,activation='relu',input_dim=2,kernel_initializer='he_normal'))
model.add(Dense(10,activation='relu',kernel_initializer='he_normal'))
model.add(Dense(10,activation='relu',kernel_initializer='he_normal'))
model.add(Dense(10,activation='relu',kernel_initializer='he_normal'))
model.add(Dense(1,activation='sigmoid'))

model.summary()
```

```python
model.get_weights()
```

```python
initial_weights = model.get_weights()
```

```python
model.set_weights(initial_weights)
```

```python
model.get_weights()
```

```python
model.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])
```

```python
history = model.fit(X,y,epochs=100,validation_split=0.2)
```

```python
from mlxtend.plotting import plot_decision_regions
plot_decision_regions(X,y.astype('int'), clf=model, legend=2)
```
